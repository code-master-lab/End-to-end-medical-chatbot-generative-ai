```python
import os
import time
import json
from dotenv import load_dotenv
from pageindex import PageIndexClient
```

```python
load_dotenv()
```

```python
PAGEINDEX_API_KEY = os.environ.get("PAGEINDEX_API_KEY")
```

```python
print("Key loaded:", PAGEINDEX_API_KEY is not None)
print("Key length:", len(PAGEINDEX_API_KEY) if PAGEINDEX_API_KEY else 0)
```

```python
pi_client = PageIndexClient(api_key=PAGEINDEX_API_KEY)
```

```python
print(pi_client)
```

```python
# Path to your 4000-page PDF, relative to the project root (we already moved here with os.chdir)
PDF_PATH = "Data2/Page_indexing_Medical_document.pdf"

# Quick safety check before uploading anything — confirms the path and filename are correct
print("File exists:", os.path.exists(PDF_PATH))
```

```python
# Double check the file is not empty this time
import os
file_size = os.path.getsize("Data2/Page_indexing_Medical_document.pdf")
print(f"File size: {file_size / (1024*1024):.2f} MB")
print(f"File exists: {os.path.exists('Data2/Page_indexing_Medical_document.pdf')}")
```

```python
# Submit the PDF to PageIndex's cloud servers for processing.
# This call returns almost immediately — it does NOT wait for the full 4000-page analysis to finish.
# We get back a doc_id, which acts like a tracking number for this specific document.

result = pi_client.submit_document(PDF_PATH)
doc_id = result["doc_id"]

print("Upload accepted. doc_id:", doc_id)
```

```python
# Check what status PageIndex has for this doc_id
doc_info = pi_client.get_document(doc_id)
print("Status:", doc_info["status"])
print("Doc ID:", doc_id)
```

```python
import json

# Fetch the tree PageIndex built from your PDF
tree_response = pi_client.get_tree(doc_id)
document_tree = tree_response["result"]

# Save tree locally
with open("Data2/medical_tree.json", "w", encoding="utf-8") as f:
    json.dump(document_tree, f, indent=2)

# Save doc_id locally
with open("Data2/medical_doc_id.json", "w", encoding="utf-8") as f:
    json.dump({"doc_id": doc_id}, f, indent=2)

print("Tree saved!")
print("Doc ID saved!")
print("Total tree nodes:", len(document_tree))
```
